```python
import requests
```

#### información obtenida desde: [Api Colombia](https://api-colombia.com/)

```python
url_base = 'https://api-colombia.com'
endpoint = '/api/v1/Department'

respuesta_api = requests.get(url_base + endpoint)
```

```python
# Confirmar que la solicitud fue correcta.
import pandas as pd
print(respuesta_api.status_code)
```

```python
## Guardar en una variable
Answ_json = respuesta_api.json()  
Departamentos  = pd.DataFrame.from_dict(Answ_json) 
```

```python
Departamentos.head()
```

```python
Departamentos.info()
```

```python
# Seleccionamos solo las columnas que contienen datos
colum_necesarias = [
    "id",
    "name",
    "description",
    "cityCapitalId",
    "municipalities",
    "surface",
    "population",
    "phonePrefix",
    "regionId",
    "countryId",
    "cityCapital",
]

Departamentos_cle = Departamentos[colum_necesarias].copy()
```

```python
# renombrar los titulos de las columnas al español para mantener una integridad linguistica en la presentación.
cambio_nombres = {
    "id": "Id",
    "name": "Departamento",
    "description": "Descripción",
    "cityCapitalId": "Id_Capital",
    "municipalities": "Total_Municipios",
    "surface": "Superficie_km2",
    "population": "Población",
    "phonePrefix": "Prefijo_Telefónico",
    "regionId": "Id_Región",
    "countryId" : "Id_país",
    "cityCapital" : "ciudad_capital", 
}

Departamentos_cle = Departamentos_cle.rename(columns=cambio_nombres)
```

```python
# 3. Eliminar filas duplicadas en caso de (especificamos la búsqueda por el Id ya que esa es la llave principal y no se repite con datos diferentes)

Departamentos_cle = Departamentos_cle.drop_duplicates(subset=["Id"])
```

```python
Departamentos_cle.info()
```

```python
Departamentos_cle
```

```python
# la columna de cuidad capitales está con bastante información que podemos individualizar
# Utilizamos el json_normalize para transformar la agrupación en columnas 
capitales = pd.json_normalize(Departamentos_cle["ciudad_capital"])

# 2. Nombrar las divisiones
capitales = capitales[["id", "name", "description"]].rename(
    columns={
        "id": "Id_Ciudad_Capital",
        "name": "Nombre_Capital",
        "description": "Descripción_Capital",
    }
)

# Unimos capitales con la base de información limpia (Departamentos_cle)
Departamentos_cle = pd.concat([Departamentos_cle, capitales], axis=1)

# Eliminamos la columna original
Departamentos_cle = Departamentos_cle.drop(columns=["ciudad_capital"])
```

```python
Departamentos_cle
```

```python
# Eliminar la fila que salio (Id_Ciudad_Capital) ya que en el formato original ya estaba en una columna inicial.
Departamentos_cle = Departamentos_cle.drop(columns=["Id_Ciudad_Capital"])
```

```python
Departamentos_cle.to_csv("Departamentos.csv", index=False)

print("Exportado corectamente")
```
