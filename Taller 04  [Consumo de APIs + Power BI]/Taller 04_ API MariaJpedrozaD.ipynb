```python
import requests
```

```python
## Obtenido de The Movie Database (TMDB).
base_url = "https://devsapihub.com"
endpoint = "/api-movies" 

respuesta_api = requests.get(base_url + endpoint)
```

```python
import pandas as pd
print(respuesta_api.status_code)
```

```python
## Almacenar la respuesta en Json en una variable
respu_json = respuesta_api.json()  
df_api = pd.DataFrame.from_dict(respu_json) 
```

```python
df_api.info()
```

```python
## Vizualizar la información
df_api.head()
```

```python
## para convertir el genre en vez de objeto en texto    
df_api['genre'] = df_api['genre'].apply(lambda x: ', '.join(x) if isinstance(x, list) else x)
```

```python
## Eliminar la colunma de la imagen de portada, ya que no la necesitan.
df_api = df_api.drop(columns=['image_url'])
```

```python
## Para eliminar posibles duplicados y guardar los cambios en una variable nueva.
df_limpio = df_api.drop_duplicates()
```

```python
# Para cambiar los nombres de las columnas de inglés a español para que e la presentación concuerde el idioma con la información.
cambio_nom = {
    'id': 'Id',
    'title': 'Título',
    'description': 'Descripción',
    'year': 'Año',
    'genre': 'Genero',
    'stars': 'Puntuación'  
}


cambio = df_limpio.rename(columns=cambio_nom)
```
