```python
import pandas as pd
import numpy as np
```

El documento base fue: Enfermedades derivadas de la ocupación en unidades de trabajo informal en bogotá; obtenido en el portal de datos abiertos de bogotá.

```python
## Utilicé el sep=";" porque es un archivo csv de excel y el encoding="latin" ya que el archivo tiene ñ.
trabajos_informales = pd.read_csv(r"C:\Users\Usuario\Downloads\obs_saludlab_enfermedaduti_.csv", sep=';', encoding='latin1')
```

```python
## Para verificar y limpiar los duplicados
dupli = trabajos_informales.duplicated () .sum()
print(f"Hay {dupli} duplicados.")
```

```python
trabajos_informales = trabajos_informales.drop_duplicates (keep="first")
trabajos_informales = trabajos_informales.reset_index(drop=True)
print("Duplicados eliminados.")
```

```python
## Revisar los nombres y cantidad de columnas, cantidad de valores y tipo de valor
trabajos_informales.info ()
```

```python
## Con el fin de eliminar filas vacías
trabajos_informales.dropna(inplace=True)
```

```python
trabajos_informales.info ()
```

### 1. ¿Cúal es la localidad con mayor porcentaje de escolaridad primaria incompleta?

```python
## Verificar las diferentes localidades
trabajos_informales["Localidad de ocurrencia del caso"].value_counts()
```

```python
## Colocamos el .size().unstack(fill_value=0)  para que corrija los valores y los organize de una forma más facil de leer.
localidad_esc = trabajos_informales.groupby(["Localidad de ocurrencia del caso", "Escolaridad"]).size().unstack(fill_value=0)
rangoprimaria = localidad_esc["Primaria incompleta"].sort_values(ascending=False)

print("LOCALIDADES PRIMARIA INCOMPLETA")
print( rangoprimaria.head(5))
```

```python
## Para graficar los primeros 5 datos
pri5_localidades = rangoprimaria.head(5)
plt.plot(pri5_localidades.index, pri5_localidades.values)
plt.xticks(rotation=45)
plt.show()
```

### 2. ¿Cúal es la ocupación informal con el nivel de ingresos más alto?

```python
# Para revisar los diferentes niveles de ingresos y saber sobre cual trabajar más adelante
trabajos_informales["Nivel de ingresos"].value_counts()
```

```python
## Para limpiar los niveles de ingresos repetidos tomados como diferentes por las formas de ingreso de la información
trabajos_informales["Nivel de ingresos"] = trabajos_informales["Nivel de ingresos"].str.strip().str.upper()
trabajos_informales["Nivel de ingresos"] = trabajos_informales["Nivel de ingresos"].replace({"NS/NR": "NS / NR",
                                                                                             "NS/NRENTRE 1 Y 2 SMMLV" : "ENTRE 1 Y 2 SMMLV"})

```

```python
## Revisar de que los valores fueron corregidos
trabajos_informales["Nivel de ingresos"].value_counts()
```

```python
ingresos_masaltos = trabajos_informales[trabajos_informales["Nivel de ingresos"] == "2 Y MÁS SMMLV"]
trabajomas_ingresos = ingresos_masaltos["Ocupación"].value_counts()

print("OCUPACIONES CON INGRESOS MÁS ALTOS")
print(trabajomas_ingresos.head(4))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

pri5_ocupaciones = trabajomas_ingresos.head(4)
plt.bar(pri5_ocupaciones.index, pri5_ocupaciones.values)
plt.xticks(rotation=45, ha="right")
plt.show()
```

### 3. ¿cuáles son los sintomas más frecuentes en los trabajos informales?

```python
sintomas_frecuentes = trabajos_informales ["Síntoma"].value_counts()
print("SÍNTOMAS MÁS FRECUENTES")
print(sintomas_frecuentes.head(5))
```

```python
pri5_sintomas = sintomas_frecuentes.head(5)

sns.barplot(x=pri5_sintomas.index, y=pri5_sintomas.values)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```
